```python
from bs4 import BeautifulSoup
import pandas as pd
import requests
```

```python
url ='https://en.wikipedia.org/wiki/List_of_epidemics_and_pandemics'
headers = {
    'User-Agent' : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/151.0.0.0 Safari/537.36 Edg/151.0.0.0'
} 
```

```python
soup = requests.get(url, headers= headers)
```

```python
webpage = BeautifulSoup(soup.text, 'html')
```

```python
target_tables = webpage.find_all('table')[2:]
```

```python
table1 = target_tables[0]
```

```python
table_headers = [header.text for header in table1.find_all('th')]
```

```python
df = pd.DataFrame(columns = table_headers)
```

```python
column_data = table1.find_all('tr')
column_data
```

```python
for row in column_data[1:-1]:
    row_data = row.find_all('td')
    individual_row_data = [data.text.strip() for data in row_data]
     
    length = len(df)   
    df.loc[length] = individual_row_data 
    
```

```python
import urllib
from sqlalchemy import create_engine

# 1. DATABASE CONFIGURATION

SERVER = r"AwesomeDad\SQLEXPRESS"
DATABASE = "EpidemicDB"
TARGET_TABLE = "EpidemicData"  # table name

connection_string = f"DRIVER={{ODBC Driver 17 for SQL Server}};SERVER={SERVER};DATABASE={DATABASE};Trusted_Connection=yes;"

# 2. CREATE THE SQLALCHEMY ENGINE

params = urllib.parse.quote_plus(connection_string)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}")


# 3. UPLOAD DATAFRAME TO SQL SERVER
 
try:
    print("Connecting to SQL Server and uploading data...")
    df.to_sql(name=TARGET_TABLE, con=engine, index=False, if_exists='replace')
    print(f"Success! Data successfully written to the '{TARGET_TABLE}' table in '{DATABASE}'.")

except Exception as e:
    print("\n[ERROR] The upload failed. Detailed error message below:")
    print(e)
```

```python
df.to_csv('epidemic_data.csv', index= False)
print("Success! DataFrame saved locally as 'epidemic_data.csv'.")
```

```python
# for the references
references= webpage.find("div", id= 'mwEjU')
ref_headers = ['ref_num', 'Reference']

rows = []

if references:
    all_references = references.find_all("li")  

    # 3. Loop through EVERY reference list item and add them to rows
    for index, item in enumerate(all_references, start=1):
        ref_text = item.text.strip()  # Get the clean text
        
        if ref_text:  # Skip if the item is blank
            rows.append({"ref_num": index, "Reference": ref_text})

#  to the DataFrame
ref_df = pd.DataFrame(rows, columns=ref_headers)

# 5.Progress report
print(f"Extracted {len(ref_df)} references.")
```

```python
# upload the references to the database
try:
    print("Connecting to SQL Server and uploading data...")
    ref_df.to_sql(name= 'ReferenceData', con=engine, index=False, if_exists='replace')
    print(f"Success! Data successfully written to the 'ReferenceData' table in '{DATABASE}'.")

except Exception as e:
    print("\n[ERROR] The upload failed. Detailed error message below:")
    print(e)
```
